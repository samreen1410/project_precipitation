# Reading Data

```python
#importing libraries
import pandas as pd
import numpy as np
import glob
import os
```

```python
#monthly flow rate data from 1925-2924
flow_monthly = pd.read_csv("monthly_flow_rate_fraser.csv", skiprows = 1)
flow_monthly.head(15)
```

```python
#monthly temperature data from 1925 - 2007
temp_monthly_1925_2007 = pd.read_csv("monthly_1925-2007.csv")
temp_monthly_1925_2007.head()
```

```python
# Get all files that start with 'daily_20'
daily_files = glob.glob("daily_20*.csv")
daily_files
```

```python
#reading and merging all daily temperature data files from 2008 - 2024
daily_list = []
for file in daily_files:
    df = pd.read_csv(file)
    daily_list.append(df)
temp_daily_2008_2024 = pd.concat(daily_list, ignore_index=True)
temp_daily_2008_2024.head()
```

```python
print("Flow monthly shape:", flow_monthly.shape)
print("Monthly temp shape:", temp_monthly_1925_2007.shape)
print("Daily temp shape:", temp_daily_2008_2024.shape)
```

# Cleaning flow data

```python
#dropping unwanted columns from flow rate data
flow_clean = flow_monthly.drop(columns=[' ID', 'TYPE', 'Mean'])
flow_clean = flow_clean.loc[:, ~flow_clean.columns.str.contains('SYM')]
flow_clean = flow_clean[(flow_clean['Year'] >= 1925) & (flow_clean['Year'] <= 2024)]
flow_clean.head()
```

```python
#changing the format of the dataset to match others
flow = flow_clean.melt(
    id_vars=['PARAM', 'Year'],
    var_name='Month',
    value_name='Discharge'
)
flow.head()
```

```python
#converting month names to numbers as in other datasets
month_dict = {
    'Jan':1, 'Feb':2, 'Mar':3, 'Apr':4,
    'May':5, 'Jun':6, 'Jul':7, 'Aug':8,
    'Sep':9, 'Oct':10, 'Nov':11, 'Dec':12
}
flow['Month'] = flow['Month'].map(month_dict)
```

# Cleaning temp data

```python
temp_daily_2008_2024.columns
```

```python
temp_monthly_1925_2007.columns
```

```python
# taking necessary columns and making sure numeric values
temp_monthly_1925_2007 = temp_monthly_1925_2007[
    ['Year','Month','Total Precip (mm)','Mean Temp (°C)']
].copy()
temp_monthly_1925_2007['Mean Temp (°C)'] = pd.to_numeric(
    temp_monthly_1925_2007['Mean Temp (°C)'],
    errors='coerce'
)
temp_monthly_1925_2007['Total Precip (mm)'] = pd.to_numeric(
    temp_monthly_1925_2007['Total Precip (mm)'],
    errors='coerce'
)
```

```python
# making sure numeric values
daily_clean = temp_daily_2008_2024.copy()
daily_clean['Mean Temp (°C)'] = pd.to_numeric(
    daily_clean['Mean Temp (°C)'],
    errors='coerce'
)
daily_clean['Total Precip (mm)'] = pd.to_numeric(
    daily_clean['Total Precip (mm)'],
    errors='coerce'
)
```

```python
# get monthly temp avergaes from daily data
temp_monthly_2008_2024 = (
    daily_clean
    .groupby(['Year','Month'])['Mean Temp (°C)']
    .mean()
    .reset_index()
)
```

```python
# get monthly prec avergaes from daily data
prec_monthly_2008_2024 = (
    daily_clean
    .groupby(['Year','Month'])['Total Precip (mm)']
    .sum()
    .reset_index()
)
```

```python
#renaming columns 
monthly_1925_2007 = temp_monthly_1925_2007[
    ['Year','Month','Mean Temp (°C)','Total Precip (mm)']
].rename(columns={
    'Mean Temp (°C)': 'Temp',
    'Total Precip (mm)': 'Prec'
})
temp_2008_2024 = temp_monthly_2008_2024.rename(
    columns={'Mean Temp (°C)': 'Temp'}
)
prec_2008_2024 = prec_monthly_2008_2024.rename(
    columns={'Total Precip (mm)': 'Prec'}
)
```

```python
#merging datasets to get ne dataset with monthly temp and prec from 
#1925-2024
climate = pd.merge(
    temp_2008_2024,
    prec_2008_2024,
    on=['Year','Month'],
    how='inner'
)
climate = pd.concat(
    [monthly_1925_2007, climate],
    ignore_index=True
)
#climate = climate.sort_values(
 #   ['Year','Month']
#).reset_index(drop=True)
print(climate.head())
print("Shape:", climate.shape)
```

```python
#thus our final cleaned 2 datasets:
print(flow.head())
print(climate.head())
```

# Missing data analysis

```python
def missing_data(df, name):
    print(f" Missing Data {name}")
    total_cells = df.size
    missing_cells = df.isna().sum().sum()
    missing_percent = (missing_cells / total_cells) * 100
    print(f"Total cells: {total_cells}")
    print(f"Missing cells: {missing_cells}")
    print(f"Missing percentage: {missing_percent:.2f}%")
```

```python
missing_data(flow, "Flow Dataset")
missing_data(climate, "Climate Dataset")
```

the missing percentage is not that significant. we can handle this. lets check which columns have the most missing values

```python
def missing_by_column(df, name):
    print(f"Missing Values by Column:{name}")
    col_missing = df.isna().sum().sort_values(ascending=False)
    col_percent = df.isna().mean().sort_values(ascending=False) * 100
    result = pd.DataFrame({
        "Missing Count": col_missing,
        "Missing Percentage": col_percent
    })
    print(result)
missing_by_column(flow, "Flow Dataset")
missing_by_column(climate, "Climate Dataset")
```

the missing data for temperature is alot.

```python
#lets try to drop it
climate = climate.dropna().reset_index(drop=True)
flow = flow.dropna().reset_index(drop=True)
```

```python
print("Flow dataset shape:", flow.shape)
print("Climate dataset shape:", climate.shape)

print("\nMissing values remaining:")
print(flow.isna().sum())
print(climate.isna().sum())
```

# RESEARCH QUESTION 1 ANALYSIS

Is there a statistically significant positive correlation between spring (Mar–May) temperature anomalies and Fraser River discharge during peak months (May–July)?

```python
#Temperature Anomaly: calculating temperature anomalies
climate['Temp_anomaly'] = climate['Temp'] - climate['Temp'].mean()
```

```python
#Spring months (May - March)
spring_temp = climate[climate['Month'].isin([3,4,5])]
spring_temp = spring_temp.groupby('Year')['Temp_anomaly'].mean().reset_index()
```

```python
#Peak Discharge (May - July)
peak_flow = flow[flow['Month'].isin([5,6,7])]
peak_flow = peak_flow.groupby('Year')['Discharge'].mean().reset_index()
```

```python
#merging above datasets to answer the question
q1 = pd.merge(spring_temp, peak_flow, on='Year')
```

```python
#removing NaNs
m = np.logical_and(~np.isnan(q1['Temp_anomaly'].values), ~np.isnan(q1['Discharge'].values))
```

```python
#correlation - Pearson correlation
import scipy.stats as stat
r, p = stat.pearsonr(q1['Temp_anomaly'].values[m], q1['Discharge'].values[m])

print("r =", np.round(r, 3))
print("p =", p)
```

```python
#scatter plot
plt.figure(figsize=(8,6))
plt.scatter(q1['Temp_anomaly'], q1['Discharge'], marker='o')
plt.xlabel('Spring Temperature Anomaly (°C)', fontsize=12)
plt.ylabel('Peak Discharge (m³/s)', fontsize=12)
plt.title('Spring Temperature Anomaly vs Fraser River Peak Discharge', fontsize=14)

# annotation
plt.annotate(
    'Pearson r: ' + str(np.round(r, 3)) + '\n p-value: ' + str(np.round(p, 5)),
    xy=(0.05, 0.85),
    xycoords='axes fraction',
    color='r'
)

plt.show()
```

r = -0.057, weak correlation and p = 0.67, not statistically significant and >>> 0.05

# RESEARCH QUESTION 2 ANALYSIS

Is winter precipitation (Dec–Feb) more strongly correlated with spring/summer discharge than precipitation during other seasons?

```python
#defining seasons
def get_season(month):
    if month in [12,1,2]:
        return 'Winter'
    elif month in [3,4,5]:
        return 'Spring'
    elif month in [6,7,8]:
        return 'Summer'
    else:
        return 'Fall'

climate['Season'] = climate['Month'].apply(get_season)
flow['Season'] = flow['Month'].apply(get_season)
```

```python
# Spring/Summer discharge (May - July)
peak_flow = flow[flow['Month'].isin([5,6,7])]
peak_flow = peak_flow.groupby('Year')['Discharge'].mean().reset_index()
```

```python
#looping through seasons and computing their correlations
results = {}
for season in ['Winter','Spring','Summer','Fall']:
    
    # seasonal precipitation
    prec = climate[climate['Season'] == season]
    prec = prec.groupby('Year')['Prec'].sum().reset_index()
    
    # merging with discharge
    df = pd.merge(prec, peak_flow, on='Year')
    
    # remove NaNs
    m = np.logical_and(~np.isnan(df['Prec']), ~np.isnan(df['Discharge']))
    
    # pearson correlation
    r, p = stat.pearsonr(df['Prec'][m], df['Discharge'][m])   
    results[season] = (r, p)

print(results)
```

```python
#plotting bar graph this time
seasons = list(results.keys())
r_vals = [results[s][0] for s in seasons]
plt.figure(figsize=(8,6))
plt.bar(seasons, r_vals)
plt.title("Seasonal Precipitation vs Peak Fraser River Discharge", fontsize=14)
plt.ylabel("Pearson r", fontsize=12)
plt.show()
```

```python
#putting results in a table
results_table = pd.DataFrame({
    'Season': list(results.keys()),
    'r': [round(results[s][0], 3) for s in results],
    'p-value': [round(results[s][1], 3) for s in results]
})

results_table
```

# RESEARCH QUESTION 3 ANALYSIS

Has the timing of peak Fraser River discharge shifted earlier in recent decades (2005–2024 vs 1958–1984)?

```python
#finding peak month each year
peak_month = flow.loc[flow.groupby('Year')['Discharge'].idxmax()]
peak_month.head()
```

```python
#splitting in 2 periods
#early: 1958 - 1984
#late: 2005 - 2024
early = peak_month[(peak_month['Year'] >= 1958) & (peak_month['Year'] <= 1984)]['Month']
late = peak_month[(peak_month['Year'] >= 2005) & (peak_month['Year'] <= 2024)]['Month']
```

```python
#performing t-test
from scipy.stats import ttest_ind

t, p = ttest_ind(early, late)
print("t =", round(t,3))
print("p =", p)
```

**p ~ 0.05: kinda significant i guess!**

```python
print("Early mean:", early.mean())
print("Late mean:", late.mean())
```

**Peak shifted is early. wasn't expecting that**

```python
# getting peak counts for May, June and July
print(early.value_counts())
print(late.value_counts())
```

```python
#trying to make a boxplot
plt.boxplot([early, late], labels=['1958–1984', '2005–2024'])
plt.ylabel("Month of Peak Discharge", fontsize=12)
plt.title("Shift in Fraser River Peak Discharge Timing", fontsize=14)
plt.show()
```

# RESEARCH QUESTION 4 ANALYSIS

Do El Niño and La Niña years show statistically significant differences in spring temperature anomalies and resulting discharge anomalies?

```python
#Reading mei data
M = pd.read_csv('MEI_data_only.txt', sep='\t', index_col=0)
print(M.head())
```

```python
#defining enso phases
enso = M.mean(axis=1).reset_index()
enso.columns = ['Year', 'MEI']
enso['Phase'] = np.where(enso['MEI'] > 0.5, 'ElNino',
                 np.where(enso['MEI'] < -0.5, 'LaNina', 'Neutral'))
enso = enso[enso['Phase'] != 'Neutral']
```

```python
#getting years of our interest
enso = enso[(enso['Year'] >= 1958) & (enso['Year'] <= 2024)]
print(enso['Phase'].value_counts())
```

```python
#Spring temp anomalies
spring_temp = climate[climate['Month'].isin([3,4,5])]
spring_temp = spring_temp.groupby('Year')['Temp_anomaly'].mean().reset_index()
```

```python
#merging
q4_temp = pd.merge(spring_temp, enso, on='Year')
```

```python
#t-test
elnino_temp = q4_temp[q4_temp['Phase'] == 'ElNino']['Temp_anomaly']
lanina_temp = q4_temp[q4_temp['Phase'] == 'LaNina']['Temp_anomaly']
t_temp, p_temp = ttest_ind(elnino_temp, lanina_temp)
print("Temp p =", p_temp)
print("El Niño mean:", elnino_temp.mean())
print("La Niña mean:", lanina_temp.mean())
```

```python
#plotting boxplot
plt.boxplot([elnino_temp, lanina_temp], tick_labels=['El Niño', 'La Niña'])
plt.title("Spring Temperature Anomalies by ENSO Phase")
plt.ylabel("Temperature Anomaly (°C)")
plt.axhline(0, color='red')
plt.show()
```

```python
#discharge anomalies
flow_annual['Anomaly'] = flow_annual['Discharge'] - flow_annual['Discharge'].mean()
```

```python
#merging with enso
q4_flow = pd.merge(flow_annual, enso, on='Year')
```

```python
#t-test
elnino_flow = q4_flow[q4_flow['Phase'] == 'ElNino']['Anomaly']
lanina_flow = q4_flow[q4_flow['Phase'] == 'LaNina']['Anomaly']

t_flow, p_flow = ttest_ind(elnino_flow, lanina_flow)
print("Flow p =", p_flow)
print("El Niño mean:", elnino_flow.mean())
print("La Niña mean:", lanina_flow.mean())
```

```python
#plotting boxplot
plt.boxplot([elnino_flow, lanina_flow], tick_labels=['El Niño', 'La Niña'])
plt.title("Discharge Anomalies by ENSO Phase")
plt.ylabel("Discharge Anomaly")
plt.axhline(0, color='red')
plt.show()
```
